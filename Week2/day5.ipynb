```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("https://s3.amazonaws.com/weclouddata/datasets/random/decision_tree/creditcard.csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
X = df.drop(['Class'] , axis=1)
y = df['Class']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test= train_test_split(X , y,  test_size=0.2, random_state=66, stratify = y)
```

```python
from collections import Counter
Counter(y_test)
```

```python
Counter(y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model = DecisionTreeClassifier()
```

```python
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score


round(accuracy_score(y_test, y_pred),3)
```

```python
conf_matrix = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=conf_matrix)
disp.plot(cmap='Blues')
plt.title('Confusion Matrix')
plt.show()
```

```python
from sklearn.metrics import classification_report
```

```python
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

Precision = $\frac{TP}{TP + FP}$
Recall = $\frac{TP}{TP + FN}$

```python
depth_list = np.array(range(1,10))
depth_list
```

```python
from sklearn.metrics import recall_score
```

```python
score_history = []
for each_depth in depth_list:
    tree_model = DecisionTreeClassifier(max_depth = each_depth)
    tree_model.fit(X_train, y_train)
    pred = tree_model.predict(X_test)
    score = recall_score(y_test , pred)
    score_history.append(score)
```

```python
plt.figure()
plt.plot(depth_list, score_history)
plt.xlabel('Tree depth')
plt.ylabel('recall score')
plt.show()
```
