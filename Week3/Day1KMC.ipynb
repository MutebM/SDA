```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
```

```python
iris = load_iris()['data']
X = iris[5:] #Row operation, coloumn operation
#Take all the row, but only the first 2 columns.
```

```python
plt.scatter(X[:, 0], X[:,1])
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=66)
labels = kmeans.fit_predict(X)
plt.scatter(X[:, 0], X[:,1], c=labels)
plt.scatter(kmeans.cluster_centers_[:,0],
            kmeans.cluster_centers_[:,1],
            marker='X', s=200 )
plt.title('KMeans')
plt.show()
```

```python
clusters = [2,4,5]

for k in clusters:
    kmeans = KMeans(n_clusters=k, random_state=47)
    labels = kmeans.fit_predict(X)
    plt.scatter(X[:, 0], X[:,1], c=labels)
    plt.scatter(kmeans.cluster_centers_[:,0],
            kmeans.cluster_centers_[:,1],
            marker='X', s=200 )
    plt.title('KMeans')
    plt.show()
```

```python
k_range = range(1,11)

inertia = []

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=66)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)
```

```python
plt.plot(k_range, inertia, marker='o')
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia / Sum of Squared Error SSE')
plt.show()
```

```python
kmeans1 = KMeans(n_clusters=3, random_state=47)
kmeans2 = KMeans(n_clusters=3, random_state=67)

labels1 = kmeans1.fit_predict(X)
labels2 = kmeans2.fit_predict(X)

plt.subplot(1 , 2, 1)
plt.scatter(X[:, 0], X[:, 1], c=labels1)
plt.title('RS47')
plt.subplot(1 , 2, 2)
plt.scatter(X[:, 0], X[:, 1], c=labels2)
plt.title('RS67')
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()['data']
X = iris[5:] #Row operation, coloumn operation
#Take all the row, but only the first 2 columns.
```

```python
scaler = StandardScaler()

# Fit the scaler to the data and transform it
scaled_data = scaler.fit_transform(X)


```

```python
kmeans_original = KMeans(n_clusters=3, random_state=66)
labels_original = kmeans_original.fit_predict(X)
kmeans_scaled = KMeans(n_clusters=3, random_state=66)
labels_scaled = kmeans_scaled.fit_predict(scaled_data)
plt.subplot(1 , 2, 1)
plt.scatter(X[:, 0], X[:, 1], c=labels_original)
plt.title('Original')
plt.subplot(1 , 2, 2)
plt.scatter(X[:, 0], X[:, 1], c=labels_scaled)
plt.title('Scaled')
```

# Silhouette Score.
Measure how well each data point fits within the assigned cluster. And how well seperated it is from other clusters.

Silhouette Score = $\frac{b_i - a_i}{max(a_i, b_i)}$

Intra-Cluster Distance ($a_i$): The average Distance between the datata point and all other point in the same cluster.

Nearest-Cluster Distance ($b_i$): The average Distance between the datata point and all other point in the nearest neighbouring cluster.

* if $a_i$ << $b_i$: The cluster is well separated
* if $a_i \approx b_i$ , Points lies between the clusters(overlap)
* $a_i$ > $b_i$, Points may be misclassified, Bad model

Silhouette Score ranges from -1 to +1
* Close to 1: Well clustered/separated
* Around 0: Overlap
* Close to -1: Very bad, Points may be misclassified

# Remember the galaxy example, Milky way - Andromeda

```python
from sklearn.metrics import silhouette_score
```

```python
unlabeled_silhouette = silhouette_score(X, labels_original)
print(f'Silhouette Score on Unlabeled Data: {unlabeled_silhouette}')
```

```python
unlabeled_silhouette_scaled = silhouette_score(scaled_data, labels_scaled)
print(f'Silhouette Score on Unlabeled Data: {unlabeled_silhouette_scaled}')
```
